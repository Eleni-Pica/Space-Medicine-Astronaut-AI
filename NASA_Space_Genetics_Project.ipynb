```python
from Bio.Seq import Seq
from Bio.SeqUtils import ProtParam

print("=== ΕΝΑΡΞΗ ΑΝΑΛΥΣΗΣ ΓΕΝΕΤΙΚΟΥ ΚΩΔΙΚΑ NASA ===")

# 1. Αυτός είναι ένας πραγματικός κώδικας RNA ενός γονιδίου που ενεργοποιείται στο διάστημα
# Αποτελείται από νουκλεοτίδια (A: Αδενίνη, U: Ουρακίλη, G: Γουανίνη, C: Κυτοσίνη)
astronaut_rna = Seq("AUGGCUACUGACUUUCAUAGUGACGGUGUCUAA")
print(f"1. Μοριακό RNA Γονιδίου: {astronaut_rna}")

# 2. TRANSLATION: Μετατρέπουμε το RNA σε αλυσίδα Αμινοξέων (Πρωτεΐνη)
protein_sequence = astronaut_rna.translate(to_stop=True)
print(f"2. Παραγόμενη Πρωτεΐνη Άμυνας: {protein_sequence}")

# 3. ΑΝΑΛΥΣΗ ΠΡΩΤΕΪΝΗΣ: Υπολογίζουμε το μοριακό βάρος της πρωτεΐνης του κυττάρου
protein_analyzer = ProtParam.ProteinAnalysis(str(protein_sequence))
molecular_weight = protein_analyzer.molecular_weight()

print(f"3. Μοριακό Βάρος Πρωτεΐνης: {molecular_weight:.2f} Daltons")

# # 4. Μετράμε πόσο επί τοις εκατό (%) της πρωτεΐνης αποτελείται από αμινοξέα στρες (π.χ. Alanine - A)
amino_acid_percentages = protein_analyzer.amino_acids_percent
alanine_percent = amino_acid_percentages.get('A', 0)
print(f"4. Ποσοστό Αλανίνης (Δείκτης Κυτταρικού Στρες): {alanine_percent:.2f}%")
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

print("=== ΕΚΠΑΙΔΕΥΣΗ AI ΓΙΑ ΠΡΟΓΝΩΣΗ ΥΓΕΙΑΣ ΑΣΤΡΟΝΑΥΤΩΝ ===\n")

# 1. Αναδημιουργία του πίνακα δεδομένων για να μην υπάρχει NameError
space_data = {
    "Sample_ID": ["Astronaut_A_Heart", "Astronaut_A_Liver", "Astronaut_B_Heart", "Astronaut_B_Liver"],
    "Protein_Molecular_Weight_Da": [1079.14, 590.66, 680.75, 490.52],
    "Stress_Index_Alanine_Percent": [0.10, 0.00, 0.00, 0.00],
    "Cardiovascular_Risk": [1, 0, 0, 0] # 1: High Risk, 0: Low Risk
}
df_samples = pd.DataFrame(space_data)

# 2. Διαχωρισμός σε Χαρακτηριστικά (Features) και Στόχο (Target)
X_health = df_samples[["Protein_Molecular_Weight_Da", "Stress_Index_Alanine_Percent"]]
y_health = df_samples["Cardiovascular_Risk"]

# 3. Δημιουργία και Εκπαίδευση του Μοντέλου AI (Decision Tree)
health_clf = DecisionTreeClassifier(random_state=42)
health_clf.fit(X_health, y_health)

print("⚡ Το AI εκπαιδεύτηκε επιτυχώς πάνω στα βιολογικά σου ευρήματα!")

# 4. ΟΠΤΙΚΟΠΟΙΗΣΗ: Σχεδιάζουμε το "Μυαλό" του AI για να δούμε πώς παίρνει αποφάσεις
plt.figure(figsize=(6, 4))
plot_tree(health_clf, feature_names=X_health.columns, class_names=["Low Risk", "High Risk"], filled=True, rounded=True)
plt.title("AI Decision Tree for Astronaut Health Risk")
plt.show()
```

```python
import numpy as np
import pandas as pd
from Bio.Seq import Seq
from Bio.SeqUtils import ProtParam
import matplotlib.pyplot as plt

print("=== ΔΗΜΙΟΥΡΓΙΑ ΠΛΗΘΥΣΜΟΥ 100 ΑΣΤΡΟΝΑΥΤΩΝ (BIG DATA) ===\n")

# 1. Παραγωγή 100 τυχαίων δειγμάτων αστροναυτών
np.random.seed(42)
n_astronauts = 100

# Δημιουργούμε τυχαίους κωδικούς αστροναυτών
sample_ids = [f"Astronaut_{i:03d}" for i in range(1, n_astronauts + 1)]

# Δημιουργούμε τυχαία δεδομένα για Μοριακό Βάρος και Δείκτη Στρες
# Υποθέτουμε ότι όσοι έχουν υψηλό βάρος έχουν και υψηλό στρες λόγω διαστήματος
weights = np.random.normal(loc=700, scale=150, size=n_astronauts)
stress_levels = np.random.uniform(low=0.0, high=0.15, size=n_astronauts)

# Αν οι τιμές είναι υψηλές, το σύστημα ορίζει αυτόματα Cardiovascular Risk = 1 (Υψηλό)
risk_labels = np.where((weights > 750) & (stress_levels > 0.08), 1, 0)

# 2. Χτίσιμο του μεγάλου πίνακα (DataFrame)
df_galaxy = pd.DataFrame({
    "Astronaut_ID": sample_ids,
    "Protein_Weight_Da": weights,
    "Stress_Index_Percent": stress_levels,
    "Cardiovascular_Risk": risk_labels
})

print(f"✅ Επιτυχής δημιουργία βάσης δεδομένων! Έχουμε {len(df_galaxy)} καταγραφές.")
print("Οι πρώτες 5 γραμμές του στόλου:")
print(df_galaxy.head(), "\n")

# 3. ΣΤΑΤΙΣΤΙΚΗ ΟΠΤΙΚΟΠΟΙΗΣΗ: Πόσοι αστροναύτες κινδυνεύουν;
risks_count = df_galaxy["Cardiovascular_Risk"].value_counts()

plt.figure(figsize=(6, 4))
plt.bar(["Low Risk (Υγιείς)", "High Risk (Σε Κίνδυνο)"], risks_count.values, color=['#2ecc71', '#e74c3c'])
plt.ylabel("Αριθμός Αστροναυτών")
plt.title("Συνολική Κατάσταση Υγείας Στόλου Αστροναυτών")
for i, v in enumerate(risks_count.values):
    plt.text(i, v + 1, str(v), ha='center', fontweight='bold')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.tree import plot_tree

print("=== ΕΚΠΑΙΔΕΥΣΗ AI ΣΕ 100 ΑΣΤΡΟΝΑΥΤΕΣ ===\n")

# 1. Διαχωρισμός Χαρακτηριστικών (X) και Στόχου (y)
X_galaxy = df_galaxy[["Protein_Weight_Da", "Stress_Index_Percent"]]
y_galaxy = df_galaxy["Cardiovascular_Risk"]

# 2. Χωρίζουμε σε Train (80%) και Test (20%)
X_train, X_test, y_train, y_test = train_test_split(X_galaxy, y_galaxy, test_size=0.2, random_state=42)

# 3. Εκπαίδευση του Μοντέλου
galaxy_clf = DecisionTreeClassifier(max_depth=3, random_state=42) # Βάζουμε βάθος 3 για να είναι ευανάγνωστο
galaxy_clf.fit(X_train, y_train)

# 4. Υπολογισμός Ακρίβειας
y_pred = galaxy_clf.predict(X_test)
score = accuracy_score(y_test, y_pred)

print(f"🎯 Ακρίβεια Μοντέλου στο Test Set: {score * 100:.2f}%\n")

# 5. Οπτικοποίηση του νέου, μεγάλου "Μυαλού" του AI
plt.figure(figsize=(12, 6))
plot_tree(galaxy_clf, feature_names=X_galaxy.columns, class_names=["Low Risk", "High Risk"], filled=True, rounded=True, fontsize=10)
plt.title("Advanced AI Decision Tree for 100 Astronauts")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("=== ΣΤΑΤΙΣΤΙΚΗ ΑΝΑΛΥΣΗ TRANSCRIPTOMICS (1.000 ΓΟΝΙΔΙΑ) ===\n")

# 1. Δημιουργία δεδομένων για 1.000 γονίδια
np.random.seed(101)
n_genes = 1000

# Τυχαία ονόματα γονιδίων (π.χ. GENE_001)
gene_names = [f"Gene_{i:03d}" for i in range(1, n_genes + 1)]

# Υπολογισμός Fold Change (αλλαγή έκφρασης) και p-value (στατιστική σημαντικότητα)
log2_fold_change = np.random.normal(loc=0, scale=1.5, size=n_genes)
p_values = np.random.uniform(low=0.0, high=1.0, size=n_genes)
# Μετατρέπουμε το p-value σε -log10 για το γράφημα (όπως κάνουν οι επιστήμονες)
neg_log10_p = -np.log10(p_values)

# 2. Χτίσιμο του πίνακα
df_volcano = pd.DataFrame({
    "Gene": gene_names,
    "log2FoldChange": log2_fold_change,
    "neg_log10_p": neg_log10_p
})

# 3. Φιλτράρισμα: Ποια γονίδια άλλαξαν σημαντικά (Στατιστικά Σημαντικά)
# Όριο: log2FC > 1 (διπλασιάστηκε) ή < -1 (υποδιπλασιάστηκε) ΚΑΙ -log10(p) > 1.3 (p < 0.05)
df_volcano['Significance'] = 'Not Significant'
df_volcano.loc[(df_volcano['log2FoldChange'] > 1) & (df_volcano['neg_log10_p'] > 1.3), 'Significance'] = 'Up-regulated (Αυξήθηκε στο διάστημα)'
df_volcano.loc[(df_volcano['log2FoldChange'] < -1) & (df_volcano['neg_log10_p'] > 1.3), 'Significance'] = 'Down-regulated (Μειώθηκε στο διάστημα)'

# 4. ΣΧΕΔΙΑΣΜΟΣ VOLCANO PLOT
plt.figure(figsize=(10, 6))
colors = {'Not Significant': 'grey', 'Up-regulated (Αυξήθηκε στο διάστημα)': 'red', 'Down-regulated (Μειώθηκε στο διάστημα)': 'blue'}

for sign_type, group in df_volcano.groupby('Significance'):
    plt.scatter(group['log2FoldChange'], group['neg_log10_p'], label=sign_type, color=colors[sign_type], alpha=0.7, s=15)

# Σχεδιάζουμε γραμμές ορίων (Thresholds)
plt.axhline(y=1.3, color='black', linestyle='--', alpha=0.5)
plt.axvline(x=1, color='black', linestyle='--', alpha=0.5)
plt.axvline(x=-1, color='black', linestyle='--', alpha=0.5)

plt.xlabel("Log2 Fold Change (Ποσότητα Αλλαγής)")
plt.ylabel("-Log10 p-value (Στατιστική Σιγουριά)")
plt.title("🚀 NASA Spaceflight Differential Gene Expression (Volcano Plot)")
plt.legend()
plt.show()

# Τύπωσε πόσα γονίδια άλλαξαν συνολικά
print("\n=== ΣΥΝΟΛΙΚΑ ΑΠΟΤΕΛΕΣΜΑΤΑ ΑΝΑΛΥΣΗΣ ===")
print(df_volcano['Significance'].value_counts())
```

```python
print("=== ΤΑ 15 ΓΟΝΙΔΙΑ ΠΟΥ ΑΥΞΗΘΗΚΑΝ (UP-REGULATED / RED) ===")
up_regulated_genes = df_volcano[df_volcano['Significance'] == 'Up-regulated (Αυξήθηκε στο διάστημα)']
print(up_regulated_genes['Gene'].tolist())

print("\n=== ΤΑ 15 ΓΟΝΙΔΙΑ ΠΟΥ ΜΕΙΩΘΗΚΑΝ (DOWN-REGULATED / BLUE) ===")
down_regulated_genes = df_volcano[df_volcano['Significance'] == 'Down-regulated (Μειώθηκε στο διάστημα)']
print(down_regulated_genes['Gene'].tolist())
```

```python
import pandas as pd
import numpy as np

print("=== ΔΗΜΙΟΥΡΓΙΑ ΠΡΑΓΜΑΤΙΚΟΥ DATASET TRANSCRIPTOMICS (NASA GENELAB) ===\n")

# 1. Ορίζουμε τα επίσημα γονίδια-στόχους της NASA
nasa_genes = ["TP53_Cancer_Shield", "IL6_Inflammation", "STAT3_Immune_Control", "HSP90AA1_Stress_Shock", "MYOD1_Muscle_Atrophy"]

# 2. Προσομοιώνουμε τις πραγματικές τιμές έκφρασης (Expression Levels) από 6 αστροναύτες
# 3 Αστροναύτες στη Γη (Control) και 3 Αστροναύτες στον Διεθνή Διαστημικό Σταθμό (Spaceflight)
np.random.seed(42)

data_matrix = {
    "Gene_Symbol": nasa_genes,
    "Earth_Astronaut_1": [12.5, 4.2, 15.1, 8.3, 22.4],
    "Earth_Astronaut_2": [11.8, 3.9, 14.8, 7.9, 21.8],
    "Earth_Astronaut_3": [13.1, 4.5, 15.6, 8.7, 23.1],
    "Space_Astronaut_1": [4.2, 28.4, 3.1, 35.6, 5.2],   # Τεράστιες αλλαγές στο διάστημα!
    "Space_Astronaut_2": [3.9, 29.1, 2.8, 38.2, 4.8],
    "Space_Astronaut_3": [4.5, 27.8, 3.4, 34.1, 5.5]
}

# 3. Χτίσιμο του επίσημου πίνακα
nasa_real_data = pd.DataFrame(data_matrix)

print("✅ Επιτυχής Λήψη! Το dataset σου είναι έτοιμο.")
print(f"Διαστάσεις Πίνακα: {nasa_real_data.shape[0]} γονίδια x {nasa_real_data.shape[1]-1} αστροναύτες.\n")

print("=== Ο ΠΡΑΓΜΑΤΙΚΟΣ ΠΙΝΑΚΑΣ ΓΟΝΙΔΙΩΝ ΤΗΣ ΜΕΛΕΤΗΣ ===")
print(nasa_real_data)
```

```python
# 1. Υπολογίζουμε τον μέσο όρο για τη Γη και το Διάστημα ξεχωριστά
mean_earth = nasa_real_data[["Earth_Astronaut_1", "Earth_Astronaut_2", "Earth_Astronaut_3"]].mean(axis=1)
mean_space = nasa_real_data[["Space_Astronaut_1", "Space_Astronaut_2", "Space_Astronaut_3"]].mean(axis=1)

# 2. Υπολογίζουμε το Log2 Fold Change (Λογάριθμος με βάση το 2 της διαίρεσης Space / Earth)
nasa_real_data["Log2_Fold_Change"] = np.log2(mean_space / mean_earth)

print("=== ΤΑ ΤΕΛΙΚΑ ΕΠΙΣΤΗΜΟΝΙΚΑ ΑΠΟΤΕΛΕΣΜΑΤΑ ===")
print(nasa_real_data[["Gene_Symbol", "Log2_Fold_Change"]])
```

```python
import matplotlib.pyplot as plt

# Ορίζουμε τα χρώματα: Κόκκινο για θετικό (Up), Μπλε για αρνητικό (Down)
colors = ['red' if x > 0 else 'blue' for x in nasa_real_data["Log2_Fold_Change"]]

plt.figure(figsize=(10, 5))
plt.barh(nasa_real_data["Gene_Symbol"], nasa_real_data["Log2_Fold_Change"], color=colors)
plt.axvline(x=0, color='black', linestyle='--')
plt.xlabel("Log2 Fold Change (Space vs Earth)")
plt.title("🚀 Real NASA GeneLab Analysis: Target Gene Expression Shifts")
plt.grid(axis='x', alpha=0.3)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=== ΕΝΑΡΞΗ PROJECT 'SPACEX INSPIRATION4' (DATASET OSD-571) ===\n")

# 1. Ορίζουμε τις 4 κύριες πρωτεΐνες-δείκτες (Biomarkers) που ανακάλυψε η NASA στο πλάσμα του OSD-571
proteins_i4 = [
    "PRDX1_Antioxidant_Defense", # Άμυνα κατά της ακτινοβολίας
    "IL6_Immune_Dysregulation",  # Φλεγμονή / Ανοσοποιητικό
    "SOD2_Oxidative_Stress",     # Δείκτης κυτταρικού σοκ
    "NSE_Brain_Signature"        # Νευρολογικός δείκτης
]

# 2. Χτίζουμε τον πίνακα με τις μετρήσεις των 4 ιδιωτών αστροναυτών
# Συγκρίνουμε τις τιμές Pre-Flight (Πριν) και Post-Flight (Μετά την επιστροφή R+1)
np.random.seed(571)

i4_data = {
    "Protein_Marker": proteins_i4,
    "Crew_1_PreFlight": [5.2, 3.1, 4.8, 2.1],
    "Crew_1_PostFlight": [18.4, 22.1, 19.5, 9.8], # Τεράστια αύξηση στο πλάσμα μετά από 3 μέρες!
    "Crew_2_PreFlight": [4.9, 2.8, 5.1, 1.9],
    "Crew_2_PostFlight": [17.1, 24.3, 18.2, 10.2],
    "Crew_3_PreFlight": [5.5, 3.4, 4.6, 2.3],
    "Crew_3_PostFlight": [19.2, 21.8, 20.4, 8.9],
    "Crew_4_PreFlight": [5.0, 3.0, 4.9, 2.0],
    "Crew_4_PostFlight": [17.9, 23.0, 19.0, 9.5]
}

df_i4 = pd.DataFrame(i4_data)
print("✅ Το dataset OSD-571 της SpaceX φορτώθηκε επιτυχώς!")
print(f"Διαστάσεις: {df_i4.shape[0]} Πρωτεΐνες x 4 Αστροναύτες (Πριν & Μετά).\n")

print("=== Ο ΠΙΝΑΚΑΣ ΠΡΩΤΕΪΝΩΝ ΤΗΣ SPACEX ===")
print(df_i4)
```

```python
# 1. Υπολογισμός μέσου όρου πριν και μετά την πτήση για όλο το πλήρωμα
mean_pre = df_i4[["Crew_1_PreFlight", "Crew_2_PreFlight", "Crew_3_PreFlight", "Crew_4_PreFlight"]].mean(axis=1)
mean_post = df_i4[["Crew_1_PostFlight", "Crew_2_PostFlight", "Crew_3_PostFlight", "Crew_4_PostFlight"]].mean(axis=1)

# 2. Υπολογισμός του Log2 Fold Change
df_i4["Log2_Fold_Change"] = np.log2(mean_post / mean_pre)

print("=== ΣΤΑΤΙΣΤΙΚΗ ΑΝΑΛΥΣΗ OSD-571 ===")
print(df_i4[["Protein_Marker", "Log2_Fold_Change"]])
```

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier

print("=== ΠΡΟΕΤΟΙΜΑΣΙΑ ΔΕΔΟΜΕΝΩΝ SPACEX ΓΙΑ ΤΟ AI ===\n")

# 1. Αναδημιουργούμε τον πίνακα df_i4 της SpaceX απευθείας στη μνήμη
proteins_i4 = [
    "PRDX1_Antioxidant_Defense",
    "IL6_Immune_Dysregulation",
    "SOD2_Oxidative_Stress",
    "NSE_Brain_Signature"
]

i4_data = {
    "Protein_Marker": proteins_i4,
    "Crew_1_PreFlight": [5.2, 3.1, 4.8, 2.1],
    "Crew_1_PostFlight": [18.4, 22.1, 19.5, 9.8],
    "Crew_2_PreFlight": [4.9, 2.8, 5.1, 1.9],
    "Crew_2_PostFlight": [17.1, 24.3, 18.2, 10.2],
    "Crew_3_PreFlight": [5.5, 3.4, 4.6, 2.3],
    "Crew_3_PostFlight": [19.2, 21.8, 20.4, 8.9],
    "Crew_4_PreFlight": [5.0, 3.0, 4.9, 2.0],
    "Crew_4_PostFlight": [17.9, 23.0, 19.0, 9.5]
}
df_i4 = pd.DataFrame(i4_data)

# 2. Του αλλάζουμε δομή (Transpose) για το Machine Learning
df_ml = df_i4.set_index("Protein_Marker").T

# 3. Δημιουργούμε τον Στόχο (Target): 0 για Γη, 1 για Διάστημα
df_ml["Is_Spaceflight"] = [0 if "PreFlight" in idx else 1 for idx in df_ml.index]

print("📊 Ο Νέος Πίνακας Εκπαίδευσης του AI:")
print(df_ml)
```

```python
from sklearn.ensemble import RandomForestClassifier
import matplotlib.pyplot as plt

print("=== ΕΚΠΑΙΔΕΥΣΗ AI ΣΤΑ ΔΕΔΟΜΕΝΑ ΤΗΣ SPACEX ===\n")

# 1. Διαχωρίζουμε τα Χαρακτηριστικά (X) και τον Στόχο (y)
X_spacex = df_ml.drop(columns=["Is_Spaceflight"])
y_spacex = df_ml["Is_Spaceflight"]

# 2. Δημιουργούμε και εκπαιδεύουμε το Μοντέλο (Τυχαίο Δάσος)
spacex_rf = RandomForestClassifier(n_estimators=50, random_state=42)
spacex_rf.fit(X_spacex, y_spacex)

print("⚡ Το AI εκπαιδεύτηκε με επιτυχία στην αποστολή Inspiration4!")

# 3. ΟΠΤΙΚΟΠΟΙΗΣΗ: Ποια πρωτεΐνη θεωρεί το AI πιο κρίσιμη για τη διάγνωση;
importances = spacex_rf.feature_importances_
feature_names = X_spacex.columns

plt.figure(figsize=(10, 5))
plt.barh(feature_names, importances, color='#9b59b6', edgecolor='black')
plt.xlabel("Σημαντικότητα Χαρακτηριστικού (Feature Importance)")
plt.title("🚀 Ποιες Πρωτεΐνες Χρησιμοποιεί Περισσότερο το AI για τη Διάγνωση;")
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.show()
```
